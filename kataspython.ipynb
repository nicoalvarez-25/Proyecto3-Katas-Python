KATA 1
Escribe una función que reciba una cadena de texto como parámetro y devuelva un diccionario con las frecuencias de cada letra en la cadena. Los espacios no deben ser considerados.

    Pasos seguidos: 
    1. Crear un diccionario vacío.
    2. Pasaremos todo a minúsculas para que no cuenten doble las mayúsculas y minúsculas.
    3. Usar un bucle for para recorrer el texto, si el caracter es un espacio lo ignoramos, si es una letra comprobamos, si existe en el diccionario sumamos 1, si no la añadimos con el valor 1. 

```python
def contar_letras(texto):
     frecuencias = {}

     texto_minusculas = texto.lower()

     for letra in texto_minusculas:
        if letra != " ":
            if letra in frecuencias:
                frecuencias[letra] = frecuencias[letra] + 1
            else:
                frecuencias[letra] = 1
     return frecuencias

    
texto_prueba = "Hola amigos"
resultado = contar_letras(texto_prueba)
print(resultado)

```

KATA 2
Dada una lista de números, obtén una nueva lista con el doble de cada valor. Usa la función map().

    Pasos seguidos:
    1. Crear una función que multiplique un sólo número por dos.
    2. Usar map() para aplicar esa pequeña función a todos los números de la lista original.
    3. Envolver el resultado con list() para que nos devuelva una lista.

```python
def mult_por_dos(numero):
    return numero * 2

def doblar_lista(lista_numeros):
    lista_resultado = list(map(mult_por_dos, lista_numeros))  #map() coge cada num y se lo pasa a la función y list() convierte el resultado en lista

    return lista_resultado

numeros_prueba = [1, 2, 3, 4, 5, 7]
resultado = doblar_lista(numeros_prueba)
print(resultado)
```

Escribe una función que tome una lista de palabras y una palabra objetivo como parámetros. La función debe devolver una lista con todas las palabras de la lista original que contengan la palabra objetivo.

    Pasos seguidos:
    1. Creamos la función con dos parámetros
    2. Crear una lista vacía para guardar los aciertos.
    3. Usamos un bucle for para analizar cada palabra de la lista, incluímos el operador in 
    

```python
def buscar_palabras(lista_palabras, palabra_objetivo):
    #creamos la lista vacía
    resultado_palabras = [] 

    #recorremos la lista que entra 
    for palabra in lista_palabras:  

        if palabra_objetivo in palabra:
            resultado_palabras.append(palabra)

    return resultado_palabras

lista_palabras = ["coche", "avión", "moto","coche deportivo", "tren", "camión", "bicicleta", "coche monovolumen"]
palabra_objetivo = "coche"

#inyectamos la lista y la palabra a buscar en el mismo orden
palabra_encontrada = buscar_palabras(lista_palabras, palabra_objetivo)

print(palabra_encontrada)
```

Kata 5
Escribe una función que tome una lista de números como parámetro y un valor opcional nota_aprobado (por defecto 5). La función debe calcular la media de los números en la lista y determinar si la media es mayor o igual que nota_aprobado. Si es así, el estado será "aprobado"; de lo contrario, "suspenso". La función debe devolver una tupla que contenga la media y el estado.

    Pasos seguidos:
    1. Creamos la función con dos parámetros, la lista de notas y la nota de aprobado que por defecto es 5.
    2. Sumamos las notas y contamos cuántas hay para calcular la media.
    3. Con un if comprobamos si la media es mayor o igual a 5 y así establecemos si es aprobado o suspenso.
    4. En el return hacemos que salga la nota media y el estado.


```python
def  evaluar_notas(lista_notas, nota_aprobado=5):
    #sumamos las notas, contamos cuántas hay y hallamos la media
    suma_notas = sum(lista_notas)
    numero_notas = len(lista_notas)   
    media = suma_notas / numero_notas

    #comprobamos si la media es mayor o igual a 5
    if media >= nota_aprobado:
        estado = "aprobado"
    else:
        estado = "suspenso"
    #devolvemos la media y el estado en una tupla
    return (media, estado)

notas_alumno = [4, 6, 4, 7, 3, 7, 5]

print(evaluar_notas(notas_alumno))
```

Kata 6
Escribe una función que calcule el factorial de un número de manera recursiva.

    Pasos seguidos:
    1. Creamos la función que recibe un número.
    2. Si el número es menor o igual que 1, devolvemos 1.
    3. Si no, devolvemos el número multiplicado por la función con ese número menos 1, es decir, llamamos a la función de manera recursiva.
    4. Devolvemos el resultado final.


```python
def calcular_factorial(numero):

    if numero <= 1:
        return 1

    else:
        return numero * calcular_factorial(numero - 1)

numero_prueba = 7
resultado = calcular_factorial(numero_prueba)

print(f"El factorial de {numero_prueba} es {resultado}")
    
```

Kata 14
Crea una función que retorne las palabras de una lista que comiencen con una letra en específico. Usa la función filter().

    Pasos seguidos:
    1. Definimos la función que selecciona la letra dentro de la función principal, así puede acceder a la letra buscada.
    2. Pasamos la palabra y la letra a minúscula para evitar fallos.
    3. .startswith comprueba si el texto empieza por esa letra.
    4. Usamos filter() para aplicar la función a la lista de palabras.
    5. Convertimos el resultado en una lista.
    6. Comprobamos que funciona.

```python
def filtrar_por_letra(lista_palabras, letra_objetivo):
    #función que selecciona la letra dentro de la función principal
    def empieza_con_letra(palabra):
        #pasamos palabra y letra a minúscula
        palabra_minuscula = palabra.lower()
        letra_minuscula = letra_objetivo.lower()
        #.startswith para ver si el texto empieza por esa letra
        if palabra_minuscula.startswith(letra_minuscula):
            return True
        else:
            return False
    #filter() para aplicar la función a la lista
    resultado_filtro = filter(empieza_con_letra, lista_palabras)
    #convertimos en lista
    return list(resultado_filtro)

palabras_prueba = ["Violín", "Saxofón", "Violonchelo", "Piano", "Guitarra"]
letra_buscada = "v"

resultado = filtrar_por_letra(palabras_prueba, letra_buscada)

print(f"Resultado: {resultado}")
```

Kata 16
Escribe una función que tome una cadena de texto y un número entero n como parámetros y devuelva una lista de todas las palabras que sean más largas que n. Usa la función filter().

    Pasos seguidos: 
    1. Creamos la función principal con los parámetros del texto y la longitud.
    2. Separamos el texto en palabras individuales con .split().
    3. Definimos la función que separa las palabras por longitud dentro de la función principal usando lambda ya que simplifica el código.
    4. Aplicamos el filter() con la función anterior y la lista de palabras.
    5. Convertimos el resultado en lista.
    6. Comprobamos que funciona.


```python
def filtrar_por_longitud(texto, n):
    #separamos el texto con .split()
    lista_palabras = texto.split()
    #función lambda para separar las palabras por su longitud
    filtrar_por_longitud = lambda palabra: len(palabra) > n
    #usamos filter() con la función anterior y la lista de palabras
    resultado_filtro = filter(filtrar_por_longitud, lista_palabras)
    #convertimos en lista
    return list(resultado_filtro)

frase_prueba = "Mi ciudad es el mejor lugar para vivir del mundo"
longitud_minima = 4
#ejecutamos con la lista y la longitud de ejemplo
resultado = filtrar_por_longitud(frase_prueba, longitud_minima)

print(f"{resultado}")
```

    Pasos seguidos:
    1. Importamos el reduce desde la librería functools de Python.
    2. Creamos la función que combina los dígitos matemáticamente, multiplicamos por 10 para desplazar las cifras a la izquierda y añadir la nueva.

```python
#creamos la función principal
def buscar_primer_duplicado(lista):
    #creamos un conjunto vacío que haga la función de "memoria" de los números por los que ya pasó el bucle
    elementos_vistos = set()
    #recorremos cada elemento de la lista
    for elemento in lista:
        #comprobamos si el elemento actual ya está en nuestra memoria
        if elemento in elementos_vistos:
            #si ya estaba es el primer duplicado y return lo expulsa
            return elemento
        #si no estaba lo guarda usando .add() y continúa
        else:
            elementos_vistos.add(elemento)
    #si no se encuentra ningún elemento duplicado no devuelve nada
    return None
#comprobamos que funcione con esta lista de números en la que se repiten dos
numeros_prueba = [1, 7, 9, 14, 7, 25, 55, 9]
resultado = buscar_primer_duplicado(numeros_prueba)

print(f"El primer elemento duplicado es: {resultado}")
```

Caso de uso:
   a. Crear dos usuarios: "Alicia" con saldo inicial de 100 y "Bob" con saldo inicial de 50, ambos con cuenta corriente.

        Pasos seguidos:
        1. Definimos la clase principal.

        3. Creamos un método para agregar dinero y mostramos con print() la cantidad que se ha ingresado.
        4. Creamos método para retirar dinero controlando los posibles errores, si el usuario no tiene una cuenta o si el saldo de su cuenta es insuficiente con if y raise, si no tienen errores retiramos el dinero.

Kata 36
Crea una función llamada procesar_texto
 - Procesa un texto según la opción especificada: contar_palabras, reemplazar_palabras o eliminar_palabra.
 - Código a seguir:
    1. Crear una función contar_palabras que cuente el número de veces que aparece cada palabra en el texto y devuelva un diccionario.
    2. Crear una función reemplazar_palabras para sustituir una palabra_original por una palabra_nueva en el texto y devolver el texto modificado.
    3. Crear una función eliminar_palabra que elimine una palabra del texto y devuelva el texto sin ella.
    4. Crear la función procesar_texto que reciba un texto, una opción ("contar", "reemplazar", "eliminar") y un número variable de argumentos según la opción elegida.

Caso de uso:
   - Verificar el funcionamiento completo de procesar_texto.


    Pasos seguidos: 
    1. Creamos una función que cuente las palabras, usamos split() para separar el texto.
    2. Con un bucle for-in recorremos todas las palabras, si la palabra ya la tenemos sumamos 1, si no, creamos el valor con 1.
    3. Creamos una función para reemplazar palabras, usamos el método replace() para reemplazarlas.
    4. Creamos una función para eliminar las palabras, usamos replace() y strip() para eliminar espacios en blanco y demás.
    5. Creamos la función principal con un número variable de argumentos(*args), en ella primero comprobamos la opción que ha elegido el usuario, si no elige ninguna de las válidas(contar, reemplazar y eliminar) provocamos el error con raise y ValueError.
    6. Creamos un texto de prueba y lo ejecutamos con las diferentes opciones de las funciones.
    

```python
#función para contar palabras
def contar_palabras(texto):
    #separamos el texto en una lista de palabras
    palabras = texto.split()
    diccionario_conteo = {}

    for palabra in palabras:
        if palabra in diccionario_conteo:
            #si ya la tenemos registrada +1
            diccionario_conteo[palabra] += 1
        else:
            #si no está registrada la registramos con el valor 1
            diccionario_conteo[palabra] = 1

    return diccionario_conteo
#función para reemplazar palabras
def reemplazar_palabras(texto, palabra_original, palabra_nueva):
    #usamos replace() para reemplazar
    return texto.replace(palabra_original, palabra_nueva)
#función para eliminar palabras
def eliminar_palabra(texto, palabra):
    #reemplazamos la palabra por un espacio en blanco y eliminamos posibles dobles espacios
    texto_limpio = texto.replace(palabra, "").replace("  ", " ").strip()
    return texto_limpio
#función princpal con número variable de argumentos(*args)
def procesar_texto(texto, opcion, *args):
    #comprobamos que opción ha elegido el usuario y llamamos a la función correspondiente
    if opcion == "contar":
        return contar_palabras(texto)

    elif opcion == "reemplazar":
        #args[0] será la palabra original y args[1] la nueva
        return reemplazar_palabras(texto, args[0], args[1])

    elif opcion == "eliminar":
        #args[0] será la palabra a borrar
        return eliminar_palabra(texto, args[0])

    else:
        raise ValueError("Opción no válida. Usa 'contar', 'reemplazar' o 'eliminar'.")

#-------- CASO DE USO --------

texto_prueba = "está empezando el invierno en mi ciudad"
#Prueba 1 - contar
resultado_contar = procesar_texto(texto_prueba, "contar")
print("Opción 'contar':")
print(resultado_contar)
#Prueba 2 - reemplazar
#pasamos 2 argumentos al final que serán la palabra reemplazada y por la que se reemplaza
resultado_reemplazar = procesar_texto(texto_prueba, "reemplazar", "invierno", "otoño")
print("Opción 'reemplazar', 'invierno' por 'otoño':")
print(resultado_reemplazar)
#Prueba 3 - eliminar
resultado_eliminar = procesar_texto(texto_prueba, "eliminar", "ciudad")
print("Opción 'eliminar', la palabra 'ciudad':")
print(resultado_eliminar)
```

    5. Estipulamos los rangos según el enunciado con if, elif y else.
    6. Con el except capturamos los inputs que estén fuera del rango estipulado.

```python
#creamos la función principal
def determinar_calificacion():
    #pedimos la nota la usuario
    entrada_usuario = input("Introduce la califiación numérica del alumno (0-100):")
    #usamos try-except para controlar los erroes
    try: 
        #pasamos el input a número decimal con float()
        nota = float(entrada_usuario)
        #controlamos que las notas solo vayan de 0 a 100 en caso contrario provocamos el error
        if nota < 0 or nota > 100:
            raise ValueError("Calificación fuera de rango")
        #establecemos los rangos de las notas
        if 0 <= nota <= 69:
            print(f"Calificación: Insuficiente")

        elif 70 <= nota <= 79:
            print(f"Calificación: Bien")

        elif 80 <= nota <= 89:
            print(f"Calificación: Muy bien")

        elif 90 <= nota <= 100:
            print(f"Calificación: Excelente")
    #controlamos si el input está fuera del rango establecido
    except ValueError as error_detectado:
        print(f"¨{error_detectado}")


determinar_calificacion()

        
```

Kata 39
Escribe una función que tome dos parámetros: figura (una cadena que puede ser "rectangulo", "circulo" o "triangulo") y datos (una tupla con los datos necesarios para calcular el área de la figura).

    Pasos seguidos:
    1. Importamos la librería matemática de Python para tener el número Pi.
    2. Creamos la función principal con dos parámetros.
    3. Pasamos el nombre de la figura a minúscula para evitar problemas.
    4. Dentro de un try-except y usando if, elif y else ponemos los parámetros y la fórmula para calcular cada una de las áreas.
    5. En el else devolvemos un mensaje si nos piden calcular el área de una figura que no tenemos programada.
    6. En el except controlamos los posibles errores de index o type y el mensaje a mostrar en ese caso.
    7. En los casos de uso mostramos que es capaz de calcular el área de las 3 figuras.


```python
#importamos la librería matemática
import math
#creamos la función principal
def calcular_area(figura, datos):
    #pasamos a minúscula
    figura = figura.lower()
    #englobamos en un try-except
    try:
        #en función de la figura aplicamos una fórmula u otra
        if figura == "rectangulo":
            base = datos[0]
            altura = datos[1]
            area = base * altura
            return area

        elif figura == "triangulo":
            base = datos[0]
            altura = datos[1]
            area = (base * altura) / 2
            return area

        elif figura == "circulo":
            radio = datos[0]
            #math.pi nos da el valor del número pi
            area = math.pi * (radio ** 2)
            return area

        else:
            #si no piden una de las figuras programadas sale este mensaje
            return(f"Error, usa 'rectangulo', 'triangulo' o 'circulo'.")
    #con estos errores controlamos los errores de posición y de tipo de dato
    except (IndexError, TypeError):
        return("Faltan datos para realizar el cálculo.")


#--------CASOS DE USO--------
#prueba 1
area_rect = calcular_area("rectangulo", (2, 4))
print(f"El área de este rectángulo es: {area_rect}")
#prueba 2
area_tri = calcular_area("triangulo", (4, 8))
print(f"El área de este triángulo es: {area_tri}")
#prueba 3
area_circ = calcular_area("circulo", (7,))
print(f"El área de este círculo es: {area_circ}")
```

Kata 40
Escribe un programa en Python que utilice condicionales para determinar el monto final de una compra en una tienda en línea, después de aplicar un descuento. El programa debe:
    a. Solicitar al usuario el precio original de un artículo.
    b. Preguntar si tiene un cupón de descuento (respuesta sí o no).
    c. Si la respuesta es sí, solicitar el valor del cupón de descuento.
    d. Aplicar el descuento al precio original, siempre que el valor del cupón sea válido (mayor a cero).
    e. Mostrar el precio final de la compra, considerando o no el descuento.
    f. Usar estructuras de control de flujo (if, elif, else) para llevar a cabo las acciones.

    Pasos seguidos:
    1. Creamos la función principal.
    2. Englobaremos el código dentro de un try-except, primero pediremos el precio del artículo al usuario mediante un input() y luego con un float() lo pasaremos a decimal.
    3. Con un if filtramos que el precio sea mayor de 0, si no, mostramos un mensaje de error. Después preguntamos con un input si existen cupones de descuento, el resultado de ese input lo pasamos a minúscula y eliminamos los posibles espacios en blanco.
    4. En otro if damos la posibilidad de poner "si"  o "sí" con un OR, luego con input() preguntamos el valor del cupón y pasamos el resultado a decimal con float().
    5. Con if comprobamos si el valor del cupón es superior a 0 y restamos directamente el valor del cupón al precio original para obtener el precio final, mostrando el resultado con print.
    6. Con else si el valor del cupón no es válido también mostramos otro mensaje.
    7. Si el usuario no tiene cupón de descuento con elif mostramos el mensaje correspondiente.
    8. Con else mostramos un mensaje si el usuario responde en la pregunta del cupón algo deferente de sí o no.
    9. Con except ValueError vemos si el usuario escribe letras en lugar de números en el precio o en el cupón de descuento.

```python
#creamos la función principal
def calcular_compra():

    try:
        #solicitamos el precio original
        entrada_precio = input("Introduce el precio original del artículo")
        #lo pasamos a decimal
        precio_original = float(entrada_precio)
        #filtramos que sea un precio válido
        if precio_original <= 0:
            return "Error, el precio del artículo debe ser mayor que 0"
        #preguntamos si el usuario tiene cupón
        tiene_cupon = input("¿Tienes un cupón de descuento? (sí/no):")
        #pasamos a minúsculas y eliminamos espacios en blanco
        tiene_cupon = tiene_cupon.lower().strip()

        if tiene_cupon == "sí" or tiene_cupon == "si":
            #preguntamos el valor del cupón
            entrada_cupon = input("Introduce el valor del cupón:")
            #pasamos a decimal
            valor_cupon = float(entrada_cupon)
            #restamos el valor del cupón al precio original si es mayor que 0
            if valor_cupon > 0:
                precio_final = precio_original - valor_cupon
                #mostrar el precio después del descuento
                print(f"¡Cupón de descuento aplicado!")
                print(f"El precio final es {precio_final}")

            else:
                #mensaje si el cupón no es válido
                print(f"El valor del cupón no es válido")
                print(f"El precio final se mantiene en {precio_original}")

        elif tiene_cupon == "no":
            #mostramos el precio sin descuento
            print(f"No se ha aplicado ningún descuento")
            print (f"El precio final es {precio_original}")

        else:
            #mostramos un mensaje si el usuario responde algo raro en la pregunta del cupón
            print(f"Respuesta no reconocida, debes contestar 'si' o 'no'")
            print(f"Operación cancelada")

    except ValueError:
        #capturamos si el usuario escribe letras en lugar de números
        print("ERROR. Por favor, introduce un valor numérico válido.")

#ejecutamos la función
calcular_compra()
```
